**Objective:**
- Use k-means clustering for image segmentation and to identify the dominant color in the image.

1. Open and display the image “dog.jpeg”. Convert the image intoa numpy array,so that itcan beused in further processing.[Hint:Use the PIL module from python]

```python
from PIL import Image
import matplotlib.pyplot as plt
import numpy as np
from sklearn.cluster import KMeans
```

```python
img = Image.open("./files/dogs.jpeg").convert("RGB")

plt.figure(figsize=(6, 6))
plt.imshow(img)
plt.title("Original Image")
plt.axis("off")
plt.show()
```

```python
img_np = np.asarray(img)
```

2. Find out the dimensions of the image and convert it intoa two-dimensional array.

```python
height, width, channels = img_np.shape
print(f"Image Dimensions: Height = {height}, Width = {width}, Channels = {channels}")
img_2d = img_np.reshape(-1, 3)
print(f"Reshaped 2D Array Shape: {img_2d.shape}")
```

 3. Use K-meansclustering with k set to 3 and cluster the image.[Hint: Refer to k-means module of scikit learn]

```python
kmeans = KMeans(n_clusters=3, random_state=42)
kmeans.fit(img_2d)
```

4. Predict the cluster label of every pixel in the image and plot it back as an image.

```python
labels = kmeans.labels_
colors = kmeans.cluster_centers_.astype(np.uint8)
quantized_pixels = colors[labels]
quantized_img_np = quantized_pixels.reshape(height, width, 3)

plt.figure(figsize=(6, 6))
plt.imshow(quantized_img_np)
plt.title("Segmented Image (K=3)")
plt.axis("off")
plt.show()
```

5. Find out the three dominant colorsin the image.[Hint: The cluster centers should correspond to three dominant colors]

```python
dominant_colors = kmeans.cluster_centers_.astype(np.uint8)
for i, color in enumerate(dominant_colors):
    print(f"Color {i+1}: R={color[0]}, G={color[1]}, B={color[2]}")

plt.figure(figsize=(6, 2))
plt.imshow([dominant_colors])
plt.title("Dominant Color Palette")
plt.axis("off")
plt.show()
```
