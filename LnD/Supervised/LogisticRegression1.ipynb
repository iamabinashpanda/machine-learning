```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler,LabelEncoder
from sklearn.metrics import accuracy_score
```

```python
df = pd.read_csv('https://raw.githubusercontent.com/shivang98/Social-Network-ads-Boost/refs/heads/master/Social_Network_Ads.csv')

df['Gender'] = LabelEncoder().fit_transform(df['Gender'])
df[['Age','EstimatedSalary']] = StandardScaler().fit_transform(df[['Age','EstimatedSalary']])

x,y = df[['Gender','Age','EstimatedSalary']],df['Purchased']
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.20,random_state=42)

model = LogisticRegression()
model.fit(x_train,y_train)
y_pred = model.predict(x_test)

accuracy_score(y_test,y_pred)*100
```

```python
sns.regplot(data=df, y='Purchased', x='Age', logistic=True)
plt.title('Age vs Purchased')
plt.show()
```

```python
# Plot EstimatedSalary vs Purchased
sns.regplot(data=df, y='Purchased', x='EstimatedSalary', logistic=True)
plt.title('Estimated Salary vs Purchased')
plt.show()
```
