```python
import pandas as pd
import matplotlib.pyplot as plt
import scipy.cluster.hierarchy as shc
```

```python
data = pd.read_csv('/content/Shopping_CustomerData.csv', usecols=['AnnualIncome','SpendingScore'])
data.head()
```

```python
plt.figure(figsize=(17,10))
plt.title('Customer Dendrogram')
plt.xlabel('Customers')
plt.ylabel('Euclidean Distance')
shc.dendrogram(shc.linkage(data, method='ward'))
plt.savefig('dendrogram.png')
plt.show()
```

```python
from sklearn.cluster import AgglomerativeClustering

cluster = AgglomerativeClustering(n_clusters=7,metric='euclidean',linkage='ward')
cluster.fit_predict(data)
```

```python
plt.figure(figsize=(15,5))
plt.scatter(y=data['AnnualIncome'],x=data['SpendingScore'],c=cluster.labels_,cmap='rainbow')
plt.show()
```

```python
pd.concat([data,pd.DataFrame(cluster.labels_)],axis=1).rename(columns={0:'Cluster'})
```
